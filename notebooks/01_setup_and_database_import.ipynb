# 01 – Brightway setup and database import

This notebook initializes the Brightway project environment and ensures
that all required background databases are available.

It performs the following steps:
1. Select or create the Brightway project
2. Run `bw2setup()` (biosphere + LCIA basics)
3. Import ecoinvent 3.8 cutoff if not already present

⚠️ **Important**
- This notebook must be executed **once** before running any others
- Due to licensing restrictions, ecoinvent data are **not included**

```python
import bw2data as bd
import bw2io as bi
from pathlib import Path
```

```python
PROJECT_NAME = "Normalization"

if PROJECT_NAME not in bd.projects:
    bd.projects.create_project(PROJECT_NAME)

bd.projects.set_current(PROJECT_NAME)
```

```python
bi.bw2setup()
```

```python
# Path to unzipped ecoinvent 3.8 cutoff ecoSpold2 datasets --> needs to be changed by every user
EI_DATASET_DIR = Path(
    r"C:\Databases\ecoinvent\ecoinvent 3.8_cutoff_ecoSpold02\datasets"
)
```

```python
EI_DB_NAME = "ecoinvent 3.8 cutoff"

if EI_DB_NAME in bd.databases:
    print(f"'{EI_DB_NAME}' already present – skipping import")
else:
    print(f"Importing {EI_DB_NAME} (this may take several minutes)…")

    importer = bi.SingleOutputEcospold2Importer(
        EI_DATASET_DIR,
        EI_DB_NAME
    )

    importer.apply_strategies()
    importer.write_database()

    print("Import completed")
```

```python
assert EI_DB_NAME in bd.databases, "ecoinvent import failed"
assert "biosphere3" in bd.databases, "biosphere database missing"

bd.databases
```
